```python
import os
import pandas as pd
from dotenv import load_dotenv, find_dotenv
```

```python
load_dotenv(find_dotenv())
root = os.getenv("ROOT_DIR")
```

```python
raw_nodes = pd.read_csv(os.path.join(root, 'data', 'raw_nodes.csv'))
raw_edges = pd.read_csv(os.path.join(root, 'data', 'raw_edges.csv'))
```

```python
raw_nodes = raw_nodes.rename(columns={'Id':'id'})
raw_edges = raw_edges.rename(columns={'id_1':'id_2'})
raw_edges = raw_edges.rename(columns={'id_0':'id_1'})



print(raw_nodes.columns)
print(raw_nodes.shape)
```

```python
print(raw_edges.columns)
print(raw_edges.head(5))
```

```python
#drop bad char and unpopular nodes
raw_nodes = raw_nodes[
    raw_nodes['name'].apply(lambda x: isinstance(x, str) and x.isascii()) 
    & (raw_nodes['popularity'] >= 20)
]
print(raw_nodes.shape)
```

```python
nodes_1 = raw_nodes.add_suffix('_1')
nodes_2 = raw_nodes.add_suffix('_2')

merged = (
    raw_edges
      .merge(nodes_1, how='inner', left_on='id_1', right_on='id_1')
      .merge(nodes_2, how='inner', left_on='id_2', right_on='id_2')
)
```

```python
print(merged.columns)
print(merged.head(5))
```

```python
merged.to_csv(os.path.join(root, 'data', 'cleaned_collaboration.csv'), index = False)
```
